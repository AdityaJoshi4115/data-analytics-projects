```python
import pandas as pd
import numpy as np 
```

```python
df =pd.read_csv("HR Employee Attrition.csv")
print("Total lenth",len(df))
print("Total Columns",len(df.columns))
df.head(10)
```

```python
# just checking the data first
print(df.shape)
print(df.isnull().sum())
print(df.columns.tolist())
```

```python
# checking attrition count
print(df['Attrition'].value_counts())

# checking full data
print(df.describe())
```

```python
# data cleaning - dropping useless columns
df = df.drop(['EmployeeCount', 'EmployeeNumber', 'Over18', 'StandardHours'], axis=1)

print("columns after dropping:", len(df.columns))
df.head()
```

```python
# converting attrition to numbers
df['Attrition'] = df['Attrition'].map({'Yes': 1, 'No': 0})

# checking attrition rate
attrition_rate = df['Attrition'].mean() * 100
print("attrition rate:", round(attrition_rate, 2), "%")
```

```python
# checking which department has most attrition
dept = df.groupby('Department')['Attrition'].mean() * 100
print("attrition by department:")
print(dept.round(2))
```

```python
# checking salary vs attrition
salary = df.groupby('Attrition')['MonthlyIncome'].mean()
print("avg salary:")
print(salary.round(2))
```

```python
# checking age vs attrition
age = df.groupby('Attrition')['Age'].mean()
print("avg age:")
print(age.round(2))
```

```python
df.to_excel('hr_analytics_cleaned.xlsx', index=False)
print("file saved!")
print("total rows exported:", len(df))
```
